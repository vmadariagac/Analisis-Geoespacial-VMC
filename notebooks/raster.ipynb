```python
import numpy as np
import matplotlib.pyplot as plt

# Tamaño del raster
n = 100

# Crear coordenadas
x = np.linspace(0, 10, n)
y = np.linspace(0, 10, n)
X, Y = np.meshgrid(x, y)

# Superficie inicial (una forma tipo colina suave)
z1 = np.sin(X) + np.cos(Y)

# Superficie final (simulamos erosión y depósito)
z2 = z1.copy()

# Simular erosión en un sector
z2[30:60, 20:50] -= 1.5

# Simular depósito en otro sector
z2[60:85, 60:90] += 1.2
```

```python
delta_h = z2 - z1
```

```python
plt.figure()
plt.imshow(delta_h)
plt.colorbar(label="Δh (m)")
plt.title("Mapa de cambio de elevación")
plt.show()
```

```python
clasificacion = np.zeros_like(delta_h)

clasificacion[delta_h < -0.5] = -1   # erosión
clasificacion[delta_h > 0.5] = 1     # depósito
```

```python
plt.figure()
plt.imshow(clasificacion)
plt.colorbar(label="Clase")
plt.title("Clasificación de cambios")
plt.show()
```
