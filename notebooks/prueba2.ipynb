```python
import pandas as pd

df = pd.read_excel("elevaciones.xlsx")
print(df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.lineplot(x=df.index, y=df["elevacion"])

plt.title("Perfil de elevaciones")
plt.xlabel("Punto")
plt.ylabel("Elevación (m)")
plt.show()
```

```python
sns.histplot(df["elevacion"], kde=True)

plt.title("Distribución de elevaciones")
plt.xlabel("Elevación (m)")
plt.show()
```

```python
# Calcular diferencia
df["delta_h"] = df["elevacion_2024"] - df["elevacion_2020"]

print(df)

# Graficar perfil de cambio
sns.lineplot(x=df.index, y=df["delta_h"])

plt.axhline(0)  # línea horizontal en cero
plt.title("Perfil de cambio altitudinal (2024 - 2020)")
plt.xlabel("Punto")
plt.ylabel("Δh (m)")
plt.show()
```

```python
print("Cambio promedio:", df["delta_h"].mean())
print("Cambio máximo:", df["delta_h"].max())
print("Cambio mínimo:", df["delta_h"].min())
```
