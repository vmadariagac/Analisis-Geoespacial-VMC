Definir una funcion que clasifique pendientes

```python
def clasificar_Pendiente(x):
    if x < 5:
        return "Plano"
    elif 5<= x < 15:
        return "Inclinado"
    elif 15<= x <= 45:
        return "Muy inclinado"
    else:
        return "Escarpado"
```

```python
Pendientes_Zona1=[20,15,8,2,45,90,76,1,8,6,23,87,52,41,3,8,65]
for p in Pendientes_Zona1:
   pendiente = clasificar_Pendiente(p)
   print(p, pendiente)
```
